```python
import sys
sys.path.append('..')

from model import AutoEncoder
from config import cfg
```

```python
def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

model = AutoEncoder(cfg.d_in)

total_params = count_parameters(model)
print(f"Trainable parameters: {total_params}")
```

```python
def get_model_size_mb(model):
    param_size = sum(p.numel() for p in model.parameters() if p.requires_grad)
    size_mb = (param_size * 4) / (1024 * 1024)  # 4 bytes per float32
    return param_size, size_mb

total_params, size_mb = get_model_size_mb(model)
print(f"Parameters: {total_params:,} | Size: {size_mb:.2f} MB")
```
